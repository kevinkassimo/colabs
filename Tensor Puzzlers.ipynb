## Puzzle 1 - ones

Compute [ones](https://numpy.org/doc/stable/reference/generated/numpy.ones.html) - the vector of all ones.

```python
def ones_spec(out):
    for i in range(len(out)):
        out[i] = 1

def ones(i: int) -> TT["i"]:
    # return ((arange(i) + 1) // (arange(i) + 1))
    # return (arange(i) % 2) + ((arange(i) + 1) % 2)
    return arange(i) * 0 + 1

test_ones = make_test("one", ones, ones_spec, add_sizes=["i"])
```

## Puzzle 2 - sum

Compute [sum](https://numpy.org/doc/stable/reference/generated/numpy.sum.html) - the sum of a vector.

```python
def sum_spec(a, out):
    out[0] = 0
    for i in range(len(a)):
        out[0] += a[i]

def sum(a: TT["i"]) -> TT[1]:
    # Note: by default, tensor scalar outputs are wrapped with tensor(...) too,
    # resulting in a 0-dimension tensor that is treated as a "scalar".
    # zero_dim_tensor[None] forces it to create a new dim to host the value.
    return (a @ ones(a.shape[0]))[None]
    # return arange(1) + (a @ ones(a.shape[0]))

test_sum = make_test("sum", sum, sum_spec)
```

```python
run_test(test_sum)
```

## Puzzle 3 - outer

Compute [outer](https://numpy.org/doc/stable/reference/generated/numpy.outer.html) - the outer product of two vectors.

```python
def outer_spec(a, b, out):
    for i in range(len(out)):
        for j in range(len(out[0])):
            out[i][j] = a[i] * b[j]

def outer(a: TT["i"], b: TT["j"]) -> TT["i", "j"]:
    # This is the "right" way to think about it. Through broadcasting
    # and Hadamard (element-wise) multiplication
    return a[:,None] * b[None,:]
    # Below I "cheated" by understanding what matrix expression of outer prod is.
    # The right way should instead be thinking about element wise mult.
    # return a[:,None] @ b[None,:]

test_outer = make_test("outer", outer, outer_spec)
```

```python
run_test(test_outer)
```

## Puzzle 4 - diag

Compute [diag](https://numpy.org/doc/stable/reference/generated/numpy.diag.html) - the diagonal vector of a square matrix.

```python
def diag_spec(a, out):
    for i in range(len(a)):
        out[i] = a[i][i]

def diag(a: TT["i", "i"]) -> TT["i"]:
    # Much simpler: tensor support multi dimension array indices, and will be
    # interpreted pair-wise.
    return a[arange(a.shape[0]), arange(a.shape[0])]
    # return ((arange(a.shape[0])[None,:] == arange(a.shape[0])[:,None]) * a) @ ones(a.shape[0])
    # return where(arange(a.shape[0])[None,:] == arange(a.shape[0])[:,None], a, arange(1)) @ ones(a.shape[0])


test_diag = make_test("diag", diag, diag_spec)
```

```python
run_test(test_diag)
```

## Puzzle 5 - eye

Compute [eye](https://numpy.org/doc/stable/reference/generated/numpy.eye.html) - the identity matrix.

```python
def eye_spec(out):
    for i in range(len(out)):
        out[i][i] = 1

def eye(j: int) -> TT["j", "j"]:
    # * 1 or + 0 are shortcuts for type conversion.
    return (arange(j)[:, None] == arange(j)[None, :]) * 1
    # return where(arange(j)[:, None] == arange(j)[None, :], arange(1) + 1, arange(1))

test_eye = make_test("eye", eye, eye_spec, add_sizes=["j"])
```

## Puzzle 6 - triu

Compute [triu](https://numpy.org/doc/stable/reference/generated/numpy.triu.html) - the upper triangular matrix.

```python
def triu_spec(out):
    for i in range(len(out)):
        for j in range(len(out)):
            if i <= j:
                out[i][j] = 1
            else:
                out[i][j] = 0

def triu(j: int) -> TT["j", "j"]:
    return (arange(j)[:, None] <= arange(j)[None, :]) + 0

test_triu = make_test("triu", triu, triu_spec, add_sizes=["j"])
```

```python
run_test(test_triu)
```

## Puzzle 7 - cumsum

Compute [cumsum](https://numpy.org/doc/stable/reference/generated/numpy.cumsum.html) - the cumulative sum.

```python
def cumsum_spec(a, out):
    total = 0
    for i in range(len(out)):
        out[i] = total + a[i]
        total += a[i]

def cumsum(a: TT["i"]) -> TT["i"]:
    return a @ triu(a.shape[0])
    # Most idiomatic way to flip 0 and 1: 1 - mask
    # If mask is boolean, then ~mask also works (won't work with int!)
    # return (a * ((1 - triu(a.shape[0])) + eye(a.shape[0]))) @ ones(a.shape[0])
    # Idea: Lower triangular + dot product
    # tril_mask = (arange(a.shape[0])[:, None] >= arange(a.shape[0])[None, :]) + 0
    # return (a * tril_mask) @ ones(a.shape[0])

test_cumsum = make_test("cumsum", cumsum, cumsum_spec)
```

```python
run_test(test_cumsum)
```

## Puzzle 8 - diff

Compute [diff](https://numpy.org/doc/stable/reference/generated/numpy.diff.html) - the running difference.

```python
def diff_spec(a, out):
    out[0] = a[0]
    for i in range(1, len(out)):
        out[i] = a[i] - a[i - 1]

def diff(a: TT["i"], i: int) -> TT["i"]:
    # Use negative indexing for shift left. Shift right is harder.
    return a - (a[arange(i) - 1] * (arange(i) > 0))
    # return (a - a[arange(i) - 1]) + (a[-1] * (arange(i) == 0))
    # return (a - a[arange(i) - 1]) * (arange(i) != 0) + (a * (arange(i) == 0))
    # return (a[(arange(i) + 1) % i] - a)[(arange(i) - 1) % i] * (arange(i) != 0) + (a * (arange(i) == 0))

test_diff = make_test("diff", diff, diff_spec, add_sizes=["i"])
```

```python
run_test(test_diff)
```

## Puzzle 9 - vstack

Compute [vstack](https://numpy.org/doc/stable/reference/generated/numpy.vstack.html) - the matrix of two vectors

```python
def vstack_spec(a, b, out):
    for i in range(len(out[0])):
        out[0][i] = a[i]
        out[1][i] = b[i]

def vstack(a: TT["i"], b: TT["i"]) -> TT[2, "i"]:
    # We need to explicitly == 1 to force into boolean.
    # Otherwise our custom where won't work!
    return where(outer(arange(2), ones(a.shape[0])) == 1, b, a)
    # Torch tensors do not support [::-1] (negative steps), unlike numpy
    # return (a[None,:] * (1 - arange(2))[:, None]) + (b[None,:] * arange(2)[:, None])


test_vstack = make_test("vstack", vstack, vstack_spec)
```

```python
run_test(test_vstack)
```

## Puzzle 10 - roll

Compute [roll](https://numpy.org/doc/stable/reference/generated/numpy.roll.html) - the vector shifted 1 circular position.

```python
def roll_spec(a, out):
    for i in range(len(out)):
        if i + 1 < len(out):
            out[i] = a[i + 1]
        else:
            out[i] = a[i + 1 - len(out)]

def roll(a: TT["i"], i: int) -> TT["i"]:
    return a[(arange(i) + 1) % i]


test_roll = make_test("roll", roll, roll_spec, add_sizes=["i"])
```

## Puzzle 11 - flip

Compute [flip](https://numpy.org/doc/stable/reference/generated/numpy.flip.html) - the reversed vector

```python
def flip_spec(a, out):
    for i in range(len(out)):
        out[i] = a[len(out) - i - 1]

def flip(a: TT["i"], i: int) -> TT["i"]:
    # Faster, just use negative index hack
    return a[-arange(i) - 1]
    # Transposed eye mult. O(n^2) so better also try something else.
    # return a @ (0 + ((i - 1 - arange(i)[:,None]) == arange(i)))


test_flip = make_test("flip", flip, flip_spec, add_sizes=["i"])
```

## Puzzle 12 - compress


Compute [compress](https://numpy.org/doc/stable/reference/generated/numpy.compress.html) - keep only masked entries (left-aligned).

```python
def compress_spec(g, v, out):
    j = 0
    for i in range(len(g)):
        if g[i]:
            out[j] = v[i]
            j += 1

def compress(g: TT["i", bool], v: TT["i"], i:int) -> TT["i"]:
    # DIFFICULT!
    # A routing matrix is needed.
    # Thinking steps:
    # 1. We have to use routing matrix as there is no easily way to do "shifts"
    #    uniformly.
    # 2. cumsum shows which index we should be routing to, yielding
    #    [0 1 1 0] -> [0 1 2 2]
    # 3. We want separate row/columns to cater to different elements, so we can
    #    use comparison to split [0 1 2 2] into [[0 1 0 0] [0 0 2 2]]
    # 4. Check against g ensures only the first occurrence of index.
    return v @ where(g[:, None], arange(i) == cumsum(g+0)[:, None] - 1, 0)
    # return (v @ ((cumsum(g+0)[:,None] == (arange(i)+1)[None,:]) * g[:, None] + 0))


test_compress = make_test("compress", compress, compress_spec, add_sizes=["i"])
```

```python
run_test(test_compress)
```

## Puzzle 13 - pad_to


Compute pad_to - eliminate or add 0s to change size of vector.

```python
def pad_to_spec(a, out):
    for i in range(min(len(out), len(a))):
        out[i] = a[i]


def pad_to(a: TT["i"], i: int, j: int) -> TT["j"]:
    # Just create routing matrix with non-explicit broadcasting of mask
    return a @ ((arange(i)[:,None] == arange(j))+0)
    # Routing matrix + explicit masking
    # mask1 = outer(arange(i)+1, ones(j)) <= (arange(j)[None,:] + 1) # Remove tril (no diag)
    # mask2 = outer(arange(i)+1, ones(j)) > (arange(j)[None,:]) # Remove triu (no diag)
    # mask = mask1 * mask2
    # return a @ (mask+0)


test_pad_to = make_test("pad_to", pad_to, pad_to_spec, add_sizes=["i", "j"])
```

```python
run_test(test_pad_to)
```

## Puzzle 14 - sequence_mask


Compute [sequence_mask](https://www.tensorflow.org/api_docs/python/tf/sequence_mask) - pad out to length per batch.

```python
def sequence_mask_spec(values, length, out):
    for i in range(len(out)):
        for j in range(len(out[0])):
            if j < length[i]:
                out[i][j] = values[i][j]
            else:
                out[i][j] = 0

def sequence_mask(values: TT["i", "j"], length: TT["i", int]) -> TT["i", "j"]:
    # Idea: mask through compare arange values against broadcasted length
    return values * (arange(values.shape[1]) < length[:,None])+0


def constraint_set_length(d):
    d["length"] = d["length"] % d["values"].shape[1]
    return d


test_sequence = make_test("sequence_mask",
    sequence_mask, sequence_mask_spec, constraint=constraint_set_length
)
```

```python
run_test(test_sequence)
```

## Puzzle 15 - bincount

Compute [bincount](https://numpy.org/doc/stable/reference/generated/numpy.bincount.html) - count number of times an entry was seen.

```python
def bincount_spec(a, out):
    for i in range(len(a)):
        out[a[i]] += 1

def bincount(a: TT["i"], j: int) -> TT["j"]:
    # This is better, using identity lookup
    return ones(a.shape[0]) @ eye(j)[a]
    # If using masks, O(n^2)
    # return ((a == arange(j)[:,None])+0) @ ones(a.shape[0])


def constraint_set_max(d):
    d["a"] = d["a"] % d["return"].shape[0]
    return d


test_bincount = make_test("bincount",
    bincount, bincount_spec, add_sizes=["j"], constraint=constraint_set_max
)
```

```python
run_test(test_bincount)
```

## Puzzle 18 - linspace

## Puzzle 20 - repeat (1d)

```python
def repeat_spec(a, d, out):
    for i in range(d[0]):
        for k in range(len(a)):
            out[i][k] = a[k]

def constraint_set(d):
    d["d"][0] = d["return"].shape[0]
    return d


def repeat(a: TT["i"], d: TT[1]) -> TT["d", "i"]:
    raise NotImplementedError

test_repeat = make_test("repeat", repeat, repeat_spec, constraint=constraint_set)
```

## Puzzle 21 - bucketize
Compute [bucketize](https://pytorch.org/docs/stable/generated/torch.bucketize.html)



```python
def bucketize_spec(v, boundaries, out):
    for i, val in enumerate(v):
        out[i] = 0
        for j in range(len(boundaries)-1):
            if val >= boundaries[j]:
                out[i] = j + 1
        if val >= boundaries[-1]:
            out[i] = len(boundaries)


def constraint_set(d):
    d["boundaries"] = np.abs(d["boundaries"]).cumsum()
    return d


def bucketize(v: TT["i"], boundaries: TT["j"]) -> TT["i"]:
    raise NotImplementedError

test_bucketize = make_test("bucketize", bucketize, bucketize_spec,
                           constraint=constraint_set)
```
